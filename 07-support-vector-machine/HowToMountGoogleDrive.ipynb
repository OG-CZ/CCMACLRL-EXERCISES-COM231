## 2. Import Google Colab Drive Module

Import the drive module from google.colab to enable Google Drive mounting functionality. This module provides the interface to connect your Colab environment with your Google Drive.

```python
# Import the drive module from google.colab
from google.colab import drive
import os
import pandas as pd
import numpy as np

print("Google Colab drive module imported successfully!")
print("Additional useful libraries imported: os, pandas, numpy")
```

## 3. Mount Google Drive

Use `drive.mount()` to mount Google Drive to the Colab environment. This will prompt you to authenticate and provide access to your Google Drive files.

**Important Notes:**
- You'll be prompted to visit a URL and get an authorization code
- Paste the authorization code when prompted
- The drive will be mounted to `/content/drive/`

```python
# Mount Google Drive
drive.mount('/content/drive')

print("Google Drive mounted successfully!")
print("Your Google Drive is now accessible at: /content/drive/MyDrive/")
```

## 4. Verify Drive Mount

Check that Google Drive has been successfully mounted by listing the contents of the mount point and verifying the connection.

```python
# Verify that Google Drive is mounted
mount_path = '/content/drive'
mydrive_path = '/content/drive/MyDrive'

# Check if mount point exists
if os.path.exists(mount_path):
    print("✅ Google Drive mount point exists")
    
    # List contents of the drive root
    print("\n📁 Contents of /content/drive/:")
    for item in os.listdir(mount_path):
        print(f"  - {item}")
    
    # Check MyDrive specifically
    if os.path.exists(mydrive_path):
        print(f"\n📁 Contents of MyDrive (showing first 10 items):")
        items = os.listdir(mydrive_path)[:10]
        for item in items:
            item_path = os.path.join(mydrive_path, item)
            item_type = "📁 Folder" if os.path.isdir(item_path) else "📄 File"
            print(f"  {item_type}: {item}")
        
        if len(os.listdir(mydrive_path)) > 10:
            print(f"  ... and {len(os.listdir(mydrive_path)) - 10} more items")
    else:
        print("❌ MyDrive folder not found")
else:
    print("❌ Google Drive not mounted properly")
```

## 5. Navigate Drive Directories

Demonstrate how to navigate through Google Drive directories using os commands and path operations. This section shows various ways to explore your Drive structure.

```python
# Navigate through Google Drive directories
import os

# Define common paths
drive_root = '/content/drive/MyDrive'

# Function to list directory contents with details
def explore_directory(path, max_items=15):
    """Explore a directory and show its contents with details"""
    if not os.path.exists(path):
        print(f"❌ Path does not exist: {path}")
        return
    
    print(f"📍 Current location: {path}")
    print(f"📊 Total items: {len(os.listdir(path))}")
    print("\n📋 Directory contents:")
    
    items = os.listdir(path)[:max_items]
    for item in items:
        item_path = os.path.join(path, item)
        if os.path.isdir(item_path):
            try:
                subitem_count = len(os.listdir(item_path))
                print(f"  📁 {item}/ ({subitem_count} items)")
            except PermissionError:
                print(f"  📁 {item}/ (access denied)")
        else:
            try:
                size = os.path.getsize(item_path)
                size_str = f"{size:,} bytes" if size < 1024 else f"{size/1024:.1f} KB" if size < 1024*1024 else f"{size/(1024*1024):.1f} MB"
                print(f"  📄 {item} ({size_str})")
            except:
                print(f"  📄 {item}")
    
    if len(os.listdir(path)) > max_items:
        print(f"  ... and {len(os.listdir(path)) - max_items} more items")

# Explore the root MyDrive directory
explore_directory(drive_root)

# Example: Create and explore a sample directory structure
sample_dir = os.path.join(drive_root, 'Colab_Sample_Directory')
if not os.path.exists(sample_dir):
    os.makedirs(sample_dir)
    print(f"\n✅ Created sample directory: {sample_dir}")

# Change working directory to your Drive (optional)
os.chdir(drive_root)
print(f"\n📍 Current working directory: {os.getcwd()}")
```
