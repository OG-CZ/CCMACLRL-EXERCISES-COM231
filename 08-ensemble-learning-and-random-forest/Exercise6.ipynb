# Removing null value

```python
df_output = df.isnull().sum().sort_values(ascending=False)
display(df_output)
```

```python
cols_to_drop = df.isnull().sum()[df.isnull().sum() > 1000].index.tolist()

df_cleaned = df.drop(columns=cols_to_drop)

print(f"Original number of columns: {df.shape[1]}")
print(f"Number of columns after dropping: {df_cleaned.shape[1]}")
display(df_cleaned.head())
```

```python
missing_values_cleaned = df_cleaned.isnull().sum().sort_values(ascending=False)
display(missing_values_cleaned[missing_values_cleaned > 0])
```

# Encoding for string data

```python
# checl categorical columns with missing values
categorical_cols_with_missing = df_cleaned.select_dtypes(include='object').columns[df_cleaned.select_dtypes(include='object').isnull().any()]

# impute missing values with mode
for col in categorical_cols_with_missing:
    mode_val = df_cleaned[col].mode()[0]
    df_cleaned[col].fillna(mode_val, inplace=True)

# check missing values after impute
print("check missing values after impute:")
display(df_cleaned[categorical_cols_with_missing].isnull().sum())

# one hot encoding on all object
df_encoded = pd.get_dummies(df_cleaned, columns=df_cleaned.select_dtypes(include='object').columns, dummy_na=False)

print()
print("after one-hot encoding:")
print(df_encoded.shape)
display(df_encoded.head())
display(df_encoded.tail())
```

## 1. Train a KNN Regressor

```python
from sklearn.neighbors import KNeighborsRegressor

score_list = {}
KNN_regressor = KNeighborsRegressor(n_neighbors=7)
KNN_regressor.fit(X_train, y_train)
knn_score = KNN_regressor.score(X_test, y_test)

score_list["KNN Regressor"] = knn_score
print(f"Score is {knn_score}")
```

- Perform cross validation

```python
# put your answer here
from sklearn.model_selection import cross_val_score
from sklearn.neighbors import KNeighborsRegressor

knn = KNeighborsRegressor(n_neighbors=5)
cv_scores = cross_val_score(knn, X_train, y_train, cv=5)

print('evaluation score',cv_scores)
print('average',cv_scores.mean())

if 'score_list' not in locals():
    score_list = {}
score_list["KNN Regressor (CV)"] = cv_scores.mean()
```

## 2. Train a SVM Regression

- Perform cross validation

```python
from sklearn.model_selection import cross_val_score
from sklearn.svm import SVR

svm = SVR()

cv_scores_svm = cross_val_score(svm, X_train, y_train, cv=5)

print("cv scores", cv_scores_svm)
print("Average cv score:", cv_scores_svm.mean())

if 'score_list' not in locals():
    score_list = {}
score_list["SVM Regressor (CV)"] = cv_scores_svm.mean()
```

## 4. Train a Random Forest Regression
